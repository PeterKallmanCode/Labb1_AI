```python
#Uppgift 3

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd
from sklearn.cluster import KMeans


df = pd.read_csv("customerdata4.csv")
point = df[["Age", "Spending Score (1-100)"]].values


k = 4
kmeans_data = KMeans(n_clusters=k).fit(point)
centroids = kmeans_data.cluster_centers_
print(centroids)

plt.scatter(point[:,0], point[:,1], c=kmeans_data.labels_.astype(float), s=50, alpha=0.5)
plt.scatter(centroids[:,0], centroids[:,1], c='red', s=40, marker='x')
plt.xlabel("Age")
plt.ylabel("Spending Score")
plt.show()


#ju större k värde desto fler grupper av kluster tvingar man fram
#Det är "bra" när man har några välseparerade grupper i detta exempel, speciellt eftersom vi bara jämför två olika saker med varandra
#Jag tror man får gå lite på magkänsla alltså när man argumenterar för sin "k" värde
#Alltså hur vill vi visa datan som handlar om Age och Spending score? Hur ser datan ut?
#Är det bra för presentationssyfte att dela in den i 2 grupper, 4 grupper, eller ingen grupp? 
#När man gör k test så förändrar inte det datan, det är samma data, fast förhållanden mellan datan förstärks
```

```python
#Uppgift 4

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd
from sklearn.cluster import KMeans


df = pd.read_csv("customerdata4.csv")
point = df[["Annual Income (10kkr)", "Spending Score (1-100)"]].values


k = 4
kmeans_data = KMeans(n_clusters=k).fit(point)
centroids = kmeans_data.cluster_centers_
print(centroids)

plt.scatter(point[:,0], point[:,1], c=kmeans_data.labels_.astype(float), s=50, alpha=0.5)
plt.scatter(centroids[:,0], centroids[:,1], c='red', s=40, marker='x')
plt.xlabel("Annual Income (10kkr)")
plt.ylabel("Spending Score")
plt.show()


#kolla den här länken han har gjort liknande: https://medium.com/@robertb909/k-means-clustering-a64f859a1074
#Tror inte vi behöver göra elbow method för att hitta bästa k, utan det kan vi babbla oss fram till
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("shopping-data.csv")

#melt gör om tabellen så att varje värde där det står kategori får ett pris, man liksom lägger om tabellen
melted = df.melt(value_vars=["Fresh", "Milk", "Grocery", "Frozen", "Detergents_Paper", "Delicassen"],
                 var_name="Category",
                 value_name="Amount")

melted.plot.scatter(x="Category", y="Amount", alpha=0.6, figsize=(10,6))
plt.title("Dot plot för olika varukategorier")
plt.show()

#sätt 50000 istället för 60000 och ta bort alla outliers, trimma den 
```
