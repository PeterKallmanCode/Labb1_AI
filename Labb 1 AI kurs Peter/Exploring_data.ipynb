```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
customer_data = pd.read_csv('customerdata4.csv')

customer_data.head()
```

```python
customer_data.describe()
print(customer_data.columns.tolist())
```

```python
# Bivariate analysis

# Scatter plot
plt.scatter(customer_data.iloc[:,2].values, customer_data.iloc[:,3].values)
plt.xlabel("Age", size=12)
plt.ylabel("Annual income", size=12)
plt.show()
```

```python
# Scatter plot 2

customer_data.plot.scatter(x="Gender", y="Annual Income (10kkr)")
plt.show()
```

```python
# Pair plot using seaborn library

sns.pairplot(data=customer_data, vars=['Age','Annual Income (10kkr)','Spending Score (1-100)'])
plt.show()
```

```python
# Correlation matrix
customer_data[['Age','Annual Income (10kkr)','Spending Score (1-100)']].corr()

sns.heatmap(customer_data[['Age','Annual Income (10kkr)','Spending Score (1-100)']].corr(), annot=True, cmap='Reds')
plt.show()
```

```python
sns.boxplot(data=customer_data, x="Annual Income (10kkr)", y="Gender")
plt.show()
```

```python
# Comparing numerical and categorical data

customer_data.groupby('Gender')['Annual Income (10kkr)'].mean()
```
