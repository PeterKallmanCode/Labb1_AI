```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("shopping-data.csv")

#Vi skapar en "melted" tabell så att varje värde ska kunna gå att plotta ut i vår scatterplot
melted = df.melt(value_vars=["Fresh", "Milk", "Grocery", "Frozen", "Detergents_Paper", "Delicassen"],
                 var_name="Category",
                 value_name="Amount")

melted.plot.scatter(x="Category", y="Amount", alpha=0.6, figsize=(10,6))
plt.title("Dot plot för olika varukategorier")
plt.show()
```

```python
#Skapa dendogram med hjälp av scipy
from scipy.cluster.hierarchy import linkage, dendrogram
x = df [["Frozen", "Fresh"]].values
z = linkage(x, method = "ward")

plt.figure(figsize=(12,6))
dendrogram(z)
plt.title("Dendrogram")
plt.xlabel("Clients")
plt.ylabel("Distance")
plt.show()
```

```python
#Skapar scatterplot som visar plottar ut grupperna "Fresh" och "Frozen"
from sklearn.cluster import AgglomerativeClustering

x = df[["Frozen", "Fresh"]]


agg = AgglomerativeClustering(n_clusters=2, linkage="ward")
labels = agg.fit_predict(x)

plt.scatter(x["Frozen"], x["Fresh"], c=labels, s=10, alpha=0.4)
plt.xlabel("Frozen")
plt.ylabel("Fresh")
plt.show()
```
