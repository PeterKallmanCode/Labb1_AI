```python
#Börjar med att importera paket för att utföra beräkningar
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
#Uppgift 1

#Läser in csv fil med hjälp av pandas paketet
customer_data = pd.read_csv('customerdata4.csv')
customer_data.describe()
```

```python
#Uppgift 2

#Skapar tre olika boxplots som jämför Gender med Age, Annual Income och Spending Score

# Age vs Gender
sns.boxplot(data=customer_data, x="Age", y="Gender")
plt.title("Age Distribution by Gender")
plt.show()


# Annual Income vs Gender
sns.boxplot(data=customer_data, x="Annual Income (10kkr)", y="Gender")
plt.title("Annual Income Distribution by Gender")
plt.show()

# Spending Score vs Gender
sns.boxplot(data=customer_data, x="Spending Score (1-100)", y="Gender")
plt.title("Spending Score Distribution by Gender")
plt.show()
```

```python
#Uppgift 2

#Skapar boxplots baserat på samma som ovan, men med annan layout på kod och visuell presentation.

import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("customerdata4.csv")
df.describe()

for col in ["Age", "Annual Income (10kkr)", "Spending Score (1-100)"]:
    df.boxplot(column=col, by="Gender")
    plt.title(f"{col} by Gender")
    plt.suptitle("")
    plt.show()
```

```python
#Uppgift 3

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd
from sklearn.cluster import KMeans


df = pd.read_csv("customerdata4.csv")
point = df[["Age", "Spending Score (1-100)"]].values

# k = 4 väljs för att det skapar fyra distinkta kluster där man enklare kan tolka skillnader. 

k = 4
kmeans_data = KMeans(n_clusters=k).fit(point)
centroids = kmeans_data.cluster_centers_
print(centroids)

plt.scatter(point[:,0], point[:,1], c=kmeans_data.labels_.astype(float), s=50, alpha=0.5)
plt.scatter(centroids[:,0], centroids[:,1], c='red', s=40, marker='x')
plt.xlabel("Age")
plt.ylabel("Spending Score")
plt.show()
```

```python
#Uppgift 4

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd
from sklearn.cluster import KMeans


df = pd.read_csv("customerdata4.csv")
point = df[["Annual Income (10kkr)", "Spending Score (1-100)"]].values


k = 5
kmeans_data = KMeans(n_clusters=k).fit(point)
centroids = kmeans_data.cluster_centers_
print(centroids)

plt.scatter(point[:,0], point[:,1], c=kmeans_data.labels_.astype(float), s=50, alpha=0.5)
plt.scatter(centroids[:,0], centroids[:,1], c='red', s=40, marker='x')
plt.xlabel("Annual Income (10kkr)")
plt.ylabel("Spending Score")
plt.show()

```
